# PDF Ingestion for RAG

This notebook compares two practical PDF loaders and prepares clean documents for chunking.

Loaders covered:
- PyPDFLoader
- PyMuPDFLoader

```python
from pathlib import Path
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter


PROJECT_ROOT = Path.cwd().resolve().parents[1]
DATA_DIR = PROJECT_ROOT / "data"
PDF_PATH = DATA_DIR / "attention.pdf"

if not PDF_PATH.exists():
    raise FileNotFoundError(f"Missing sample PDF: {PDF_PATH}")

print("PDF path:", PDF_PATH)
```

```python
pypdf_loader = PyPDFLoader(str(PDF_PATH))
pypdf_docs = pypdf_loader.load()

print("PyPDFLoader")
print("pages:", len(pypdf_docs))
print("first page preview:", pypdf_docs[0].page_content[:160], "...")
print("first page metadata:", pypdf_docs[0].metadata)
```

```python
pymupdf_loader = PyMuPDFLoader(str(PDF_PATH))
pymupdf_docs = pymupdf_loader.load()

print("\nPyMuPDFLoader")
print("pages:", len(pymupdf_docs))
print("first page preview:", pymupdf_docs[0].page_content[:160], "...")
print("first page metadata keys:", list(pymupdf_docs[0].metadata.keys())[:10])
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150,
)
pdf_chunks = splitter.split_documents(pymupdf_docs)

print("\nChunking result")
print("chunks:", len(pdf_chunks))
print("first chunk chars:", len(pdf_chunks[0].page_content))
print("first chunk source:", pdf_chunks[0].metadata.get("source"))
```
