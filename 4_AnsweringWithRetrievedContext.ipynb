# 4. Answering With Retrieved Context

This notebook shows the handoff from retrieval to answer generation. The retrieved chunks become the context the model reads before it writes the final answer.

Use it to see how a question, retrieved text, and a prompt fit together before the `stuff` document-combination pattern in the next notebook.

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_classic.chains.combine_documents import create_stuff_documents_chain

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "Answer only from the provided context. If the context is incomplete, say what is missing."),
        ("human", "Context:\n{context}\n\nQuestion:\n{input}"),
    ]
)
prompt
```

## What to observe

- The prompt keeps the answer grounded in retrieved text instead of relying on memory.
- The `context` placeholder is where retrieved chunks are inserted later.
- The `input` placeholder is the user question.
- This is the simplest version of the answer-construction step that later notebooks reuse inside full RAG chains.
